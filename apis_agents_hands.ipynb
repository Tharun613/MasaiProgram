### Task 1

```python
import requests
import json
import pandas as pd

try:
    response = requests.get("https://restcountries.com/v3.1/all?fields=name,population,region,area")
    print(f"The status code of the request is: {response.status_code}")
    countries = response.json()
    print(f"Total number of countries received: {len(countries)}")
except Exception as e:
    print(f"Failed to fetch the list of countries .. {str(e)}")
    countries = []

```

### Task 2

```python
parsed_countries = [ 
    {
        "Name": country["name"]["common"],
        "Region": country["region"],
        "Area": country["area"],
        "Population": country["population"]
    }
    for country in countries
]
df = pd.DataFrame(parsed_countries)

df.head(10)
```

### Task 3

```python
largest_population_country=df.sort_values(by="Population", ascending=False).reset_index().loc[0, "Name"]

print(f"{str(largest_population_country)} has the largest population.")

largest_area_country=df.sort_values(by="Area", ascending=False).reset_index().loc[0, "Name"]
print(f"{str(largest_area_country)} has the largest area.")
```
